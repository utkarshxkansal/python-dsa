```python
arr = [10,20,55,91,1,4,5,8]
print("orignal array",arr)
n = len(arr)
for i in range(n-1,0,-1):
    for j in range(0,i):
        if arr[j+1]<arr[j]:
            arr[j+1],arr[j]=arr[j],arr[j+1]
        
   
            
        
    
print("sorted array",arr)


# bubble sort
        
        
```

```python
arr = [10,20,55,91,1,4,5,8]
print("orignal array",arr)
n = len(arr)
for i in range(0,n):
    j= i
    while(j>0 and arr[j-1]>arr[j]):
        arr[j-1],arr[j]=arr[j],arr[j-1]
        j=j-1

print("sorted array",arr)

# insertion sort

```
