```python
# hashing of no arrays
n = int(input("enter the no of ewlemets u want in ur array"))
arr = [None]*n
for i in range(0,n):
    arr[i] = int(input('enter the element'))

# precompute
hash = [0]*(10**7+1)
for i in range(0,n):
    hash[arr[i]] +=1

q = int(input("Enter the no of queriries"))
while(q >0):
    no = int(input("the no u want to know how many tuimes it repeats"))
    # //fetch
    print(hash[no])
    q = q-1
```

```python
# using dictionary

n = int(input("enter the no of ewlemets u want in ur array"))
arr = [None]*n
for i in range(0,n):
    arr[i] = int(input('enter the element'))

# precompute 
hash = {}
for i in range(0,len(arr)):
    if arr[i] in hash:
        hash[arr[i]] +=1
    else:
        hash[arr[i]] = 1


```
