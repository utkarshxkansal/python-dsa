```python
class Node:
    def __init__(self,val,left=None,right=None):
        self.val = val
        self.left = left
        self.right = right

        
        
    

class Solution:
    def all_traversals(self,root):
        tree = []
        preorder = []
        inorder = []
        postorder = []

        if(root == None):
            return

        tree.append((root,1))
        
        while tree:
            x = tree.pop()
            node = x[0]
            state = x[1]

            if(state == 1):
                preorder.append(node.val)
                tree.append((node,2))

                if(node.left):                    
                    tree.append((node.left,1))
                    
            elif(state == 2):
                inorder.append(node.val)
                tree.append((node,3))

                if(node.right):
                    tree.append((node.right,1))

            if(state == 3):
                postorder.append(node.val)
        return preorder, inorder, postorder
                
                
                

        

        
        
    
```

```python
obj = Solution()

pre, ino, post = obj.all_traversals(root)

print("Preorder:", pre)
print("Inorder:", ino)
print("Postorder:", post)
```
