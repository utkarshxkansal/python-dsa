```python
class node:
    def __init__(self,Data):
        self.data = Data
        self.next = None
    
arr = [31,34,45,512,45,12,524,1]
head = node(arr[0])
mover = head
temp = None
print(mover.data,end="->")
for i in range(1,len(arr)):
    temp = node(arr[i])
    mover.next = temp
    mover = temp
    print(mover.data,end= "->")
```

```python
def insert_head(head,data):
    new = node(data)
    new.next = head
    return new
```

```python
head = insert_head(head,66)
```

```python
mover = head
while(mover !=None):
    print(mover.data,end="->")
    mover = mover.next
```
