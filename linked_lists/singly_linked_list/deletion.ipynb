```python
# delete head
class node:
    def __init__(self,Data):
        self.data = Data
        self.next = None
```

```python
arr = [1,23,3,4,4,4,32,2,1]
head = node(arr[0])
mover = head
for i in range(1,len(arr)):
    temp = node(arr[i])
    mover.next = temp
    mover = temp
# creating a linked list 
```

```python
def remove_kth(head, k):
    previous = None
    mover3 = head

    if head is None:
        return None

    if k == 1:
        return head.next

    count = 1

    while mover3 is not None and count < k:
        previous = mover3
        mover3 = mover3.next
        count += 1

    if mover3 is None:
        return head

    previous.next = mover3.next
    return head
```
