```python
class node:
    def __init__(self,Data):
        self.data = Data
        self.next = None
    
arr = [31,34,45,512,45,12,524,1]
head = node(arr[0])
mover = head
temp = None
print(mover.data,end="->")
for i in range(1,len(arr)):
    temp = node(arr[i])
    mover.next = temp
    mover = temp
    print(mover.data,end= "->")
k = 3
```

```python
def insertion_at_k(head, k, data):
    newnode = node(data)

    if head is None:
        if k == 1:
            return newnode
        return None

    if k == 1:
        newnode.next = head
        return newnode

    cnt = 1
    temp = head

    while temp is not None and cnt < k - 1:
        temp = temp.next
        cnt += 1

    if temp is None:
        return head

    newnode.next = temp.next
    temp.next = newnode

    return head
```

```python
head = insertion_at_k(head,3,10000000000)
```

```python
mover = head
while mover is not None:
    print(mover.data, end="->")
    mover = mover.next
```
