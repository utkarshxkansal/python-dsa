```python
class node:
    def __init__(self,Data):
        self.data = Data
        self.next = None
        
```

```python
arr = [1,2,34,3,4,2,35,65,62,3,4]
head = node(arr[0])
mover = head
for i in range(1,len(arr)):
    mover.next = node(arr[i])
    mover = mover.next
```

```python
def remove_by_val(head,m):
    if(head == None): return None
    if(head.data==m):
        temp = head
        head = head.next
        return head
    mover1 = head
    previous = None
    while(mover1 != None):
        if(mover1.data == m):
            previous.next = mover1.next
        previous = mover1
        mover1 = mover1.next
    return head
        
```
