```python
class node:
    def __init__(self,Data=None,Next = None):
        self.data = Data
        self.next = Next                                
```

```python
class stack:
    def __init__(self):
        self.top = None

    def push(self,item):
        self.top = node(item,self.top)
    def pop(self):
        if self.top == None:
            print("Underflow:stack is empty")
            return None
        self.temp = self.top
        self.top = self.top.next
        self.temp = None
    def peek(self):
        if self.top == None:
            print("Stack is empty")
            return None
        print(self.top.data)
```

```python
s = stack()
s.push(5)
s.pop()
```
