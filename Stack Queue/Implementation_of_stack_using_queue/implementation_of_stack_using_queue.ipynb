```python
from collections import deque
```

```python
class stack:
    def __init__(self):
        self.q = deque()
    def push(self,x):
        self.q.append(x)#enqueue

        for i in range(1,len(self.q)):
            self.q.append(self.q.popleft())

    def pop(self):
        if not self.q:
            return None
        else:
            return self.q.popleft()

    def top(self):
        if not self.q:
            return None
        else:
            return self.q[0]

    def is_empty(self):
        if not self.q:
            return True
        else:
            return False

    def display(self):
        print(self.q)



        
        
```

```python
stq = stack()
stq.push(10)
stq.push(10)
```

```python
stq.pop()
```
