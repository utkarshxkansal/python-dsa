```python
from math import sqrt

n = int(input("Enter a no "))
factors1 = []

num = int(sqrt(n))

for i in range(1, num + 1):
    if n % i == 0:
        factors1.append(i)
        r = n // i

        if r != i:
            factors1.append(r)


factors1.sort()
print(factors1)
```
