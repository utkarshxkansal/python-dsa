```python
class node:
    def __init__(self,Data,Next = None,Back=None):
        self.data = Data
        self.next = Next
        self.back = Back

arr = [21,34,3245,56,4712,34,33,546,6752,1]
head = node(arr[0])

mover = head
temp = 0
for i in range(1,len(arr)):
    temp = node(arr[i])
    mover.next = temp    
    temp.back =  mover 
    mover = temp

```

```python
def del_kth(head,k):
    cnt = 0
    mover = head
    forward = None 
    previous = None
    while(mover != None and cnt < k-1):
        cnt +=1
        mover = mover.next
    if(mover == None or k<1):
        return head

    mover.next = forward
    mover.back = previous
    
    if(mover.next == None and mover.back == None):
        return None
    elif(mover.next == None):
        previous.next = None
        mover.back = None
                
    elif(mover.back == None):
        head = forward
        forward.back = None
        mover.next = None
    else:    
        previous.next = forward
        forward.back = previous
        mover.back = None
        mover.next = None
    return head

                
            
```
