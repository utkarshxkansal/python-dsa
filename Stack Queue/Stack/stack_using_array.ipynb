```python
class stack:
    def __init__(self):
        self.arr = []
        self.top = -1

    def push(self,x):
        self.arr.append(x)
        self.top +=1

    def pop(self):
        if self.top == -1:
            print("Stack Underflow")
            return None
        self.top -=1
        return self.arr.pop()
    def peek(self):
        if self.top ==-1:
            return None
        print(self.arr[self.top])
            
    
```

```python
st = stack()
```

```python
st.push(45)
```

```python
st.pop()
```
