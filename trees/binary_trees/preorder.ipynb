```python
class node:
    def __init__(self,data,left=None,right=None):
        self.data = data
        self.left = left
        self.right = right

root = node(1)
root.left = node(2)
root.right = node(3)
root.left.left = node(4)
```

```python
def preorder(node):
    if(node== None):
        return
    print(node.data)
    preorder(node.left)
    preorder(node.right)

    
```

```python
preorder(root)
```
