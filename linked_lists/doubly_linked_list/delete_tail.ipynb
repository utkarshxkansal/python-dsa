```python
class node:
    def __init__(self,Data,Next = None,Back=None):
        self.data = Data
        self.next = Next
        self.back = Back

arr = [21,34,3245,56,4712,34,33,546,6752,1]
head = node(arr[0])

mover = head
temp = 0
for i in range(1,len(arr)):
    temp = node(arr[i])
    mover.next = temp    
    temp.back =  mover 
    mover = temp
```

```python
def delt_tail(head):
    if head == None or head.next == None:
        return None

    tail = head

    while tail.next != None:
        tail = tail.next  # Move tail forward

    newtail = tail.back
    newtail.next = None
    tail.back = None

    return head
```

```python
head = delt_tail(head)
```
