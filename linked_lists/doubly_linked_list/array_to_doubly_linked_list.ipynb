```python
class node:
    def __init__(self,Data,Next = None,Back=None):
        self.data = Data
        self.next = Next
        self.back = Back

arr = [21,34,3245,56,4712,34,33,546,6752,1]
head = node(arr[0])
```

```python
def making_dll(head):
    mover = head
    temp = 0
    for i in range(1,len(arr)):
        temp = node(arr[i])
        mover.next = temp    
        temp.back =  mover 
        mover = temp
```

```python
mover1 = head
while(mover1 != None):
    print(mover1.data,end="->")
    mover1 = mover1.next
```
