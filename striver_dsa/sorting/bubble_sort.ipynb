```python
arr = [1, 3, 9, 11, 56, 99, 311]
n = len(arr)
didswap = 0
print("unsorted array",arr)


for i in range(n-1,0,-1):
    for j in range(0,i):
        if arr[j+1]<arr[j]:
            arr[j+1],arr[j]=arr[j],arr[j+1]
            didswap =1
print("sorted array",arr)
if didswap==1:
    print("time complexity =N")
else:
    print("its n square")
            
        
```
